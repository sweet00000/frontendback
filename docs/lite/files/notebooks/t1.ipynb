# MVP notebook
This notebook runs in Pyodide via JupyterLite.

```
# Enable requests via browser fetch (Pyodide)
import piplite, pyodide_http
await piplite.install(["pyodide-http", "requests"])
pyodide_http.patch_all()  # make requests use browser fetch; CORS rules apply

import requests
resp = await requests.get("https://jsonplaceholder.typicode.com/todos/1")
print(resp.status_code, resp.json())
```

```
import js, asyncio
est = await js.navigator.storage.estimate()
print('Storage usage bytes:', int(est.usage))
print('Storage quota bytes:', int(est.quota))
print('WebGPU available:', bool(getattr(js.navigator, 'gpu', None)))
print('User agent:', js.navigator.userAgent)
```
